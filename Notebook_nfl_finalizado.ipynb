### Tratamento de valores nulos

Nesta etapa vamos tratar os valores nulos. Na maioria das colunas do dataset o nulo não é erro de coleta, ele aparece porque aquela lance não teve aquela jogada.exemplo: em uma corrida a coluna Passer fica vazia, e FieldGoalDistance só tem valor quando a jogada é um field goal.

Por isso decidi não excluir as linhas por que se não iriamos excluir quase todo o dataset e então decidi tratar os valores das colunas para diminuir o numero de nulos 

Primeiro contei quantos valores nulos tem em cada coluna e fiz um calculo com a porcentagem de valores nulos

```python
# quantidade e porcentagem de nulos só das colunas que tem algum valor nulo

qtd_nulos = df.isna().sum()
pct_nulos = (qtd_nulos / len(df) * 100).round(2)

resumo_nulos = pd.DataFrame({"qtd": qtd_nulos, "porcentagem": pct_nulos})
resumo_nulos = resumo_nulos[resumo_nulos["qtd"] > 0].sort_values("porcentagem", ascending=False)
resumo_nulos
```

```python
# Separei em tres grupos pra ficar mais fácil de decidir o que fazer com cada coluna, baseado na porcentagem de valores nulos
muitos_nulos = resumo_nulos[resumo_nulos["porcentagem"] > 50].index.tolist()
medios_nulos = resumo_nulos[(resumo_nulos["porcentagem"] > 10) & (resumo_nulos["porcentagem"] <= 50)].index.tolist()
poucos_nulos = resumo_nulos[resumo_nulos["porcentagem"] <= 10].index.tolist()

print("Mais de 50%:", muitos_nulos)
print("Entre 10% e 50%:", medios_nulos)
print("Até 10%:", poucos_nulos)
```

### Tentativa de tratamento da Coluna BlokingPlayer

Nesse trecho eu quis descobrir onde a coluna BlockingPlayer estava vazia. Primeiro filtrei o dataframe só com as linhas em que ela é nula e guardei isso em nulos_bp. Depois dei print  do total de nulos e o percentual em relação ao dataset inteiro, para ter noção do tamanho do problema. No fim mostrei as primeiras linhas com PlayType, desc e BlockingPlayer, para olhar o tipo de jogada e a descrição de cada uma e entender por que o jogador estava vazio. Vi que quase todos os nulos, cerca de 99,97%, são de jogadas normais como kickoffs, passes e corridas, em que ninguém bloqueia um chute, então esse vazio é esperado e não é erro de coleta.

```python
# linhas onde BlockingPlayer está nulo
nulos_bp = df[df["BlockingPlayer"].isna()]

print("Total de nulos:", len(nulos_bp))
print("Percentual:", round(df["BlockingPlayer"].isna().mean() * 100, 2), "%")

nulos_bp[["PlayType", "desc", "BlockingPlayer"]].head()
```

```python
# jogadas bloqueadas na descrição, mas sem jogador na coluna
#Aqui eu vi se na descrição tem alguma descrição de  bloqueio, mas a coluna BlockingPlayer está nula, o que pode indicar um problema de registro de dados
sem_jogador = df["BlockingPlayer"].isna() & df["desc"].str.contains("BLOCKED", case=False, na=False)

# Dei um print em todos os valores que encontrei
print("Bloqueios sem jogador:", sem_jogador.sum())
df.loc[sem_jogador, "desc"].head()
```

```python
# Aqui eu tentei extrair o nome do jogador que bloqueou a jogada da coluna "desc" (narração)
# Essa expressão regular busca o texto "BLOCKED " (?:by )?\(?([A-Z]\.[A-Za-z.'\-]+) e captura o nome do jogador no formato de inicial e sobrenome quando o valor for nulo

nome = df["desc"].str.extract(r"BLOCKED (?:by )?\(?([A-Z]\.[A-Za-z.'\-]+)", expand=False)
df["BlockingPlayer"] = df["BlockingPlayer"].fillna(nome)
```

```python
# Retorna o total de valores nulos
#Da pra ver que diminuiu 306 valores o que significa que conseguimos preencher alguns valores nulos na coluna "BlockingPlayer" a partir da descrição da jogada funcionou
df["BlockingPlayer"].isna().sum()
```

#### Jogadores e descrição da jogada

Nas colunas de jogadores o nulo quer dizer que nenhum jogador fez aquele tipo de jogada, troquei o valor nulo por "Nenhum". Já o nulo da descrição esta relacionado com uma informação quenão foi registrada, por isso vai receber "sem descrição"

```python
col_jogadores = ['Passer', 'Passer_ID', 'Rusher', 'Rusher_ID', 'Receiver', 'Receiver_ID',
                 'Returner', 'Interceptor', 'Tackler1', 'Tackler2', 'PenalizedPlayer',
                 'RecFumbPlayer', 'BlockingPlayer']

df[col_jogadores] = df[col_jogadores].fillna("Nenhum")
df["desc"] = df["desc"].fillna("Sem descrição")

# conferindo se sobrou algum nulo nessas colunas
df[col_jogadores + ["desc"]].isna().sum().sum()
```

#### Colunas categóricas

Em coluna categorica estava dando erro ao usar o "fillna" com um valor que ainda não existe nas categorias. Então pesquisei e obtei por criar uma função que adiciona a categoria nova antes de preencher.

- Colunas de times "posteam" e "DefensiveTeam": Ficam "Sem posse", pois ficam nulas quando nenhum time está com a bola
- As outras colunas  como resultado do punt, tipo de penalidade, direção do passe: ficam como "Nenhum", pois o evento não aconteceu

```python
def preenche_categoria(coluna, valor):
    # Vaos adicionar a categoria nova  antes do fillna para não dar erro
    if valor not in df[coluna].cat.categories:
        df[coluna] = df[coluna].cat.add_categories(valor)
    df[coluna] = df[coluna].fillna(valor)


col_times = ['posteam', 'DefensiveTeam']

col_eventos = ['DefTwoPoint', 'TwoPointConv', 'ChalReplayResult', 'RecFumbTeam', 'FieldGoalResult',
               'ExPointResult', 'PenaltyType', 'Timeout_Team', 'ReturnResult', 'PuntResult',
               'PenalizedTeam', 'RunGap', 'RunLocation', 'PassLocation', 'PassLength',
               'PassOutcome', 'SideofField']

for coluna in col_times:
    preenche_categoria(coluna, "Sem posse")

for coluna in col_eventos:
    preenche_categoria(coluna, "Nenhum")

df[col_times + col_eventos].isna().sum().sum()
```

#### Algumas colunas  o zero tem significado

-  Na coluna"down": não existe down em kickoff e outras jogadas, então 0 quer dizer "sem down" naquela jogada
- Na coluna "FirstDown" e "GoalToGo`: são indicadores de 0 ou 1, nulo é quando não acontceu 
- Na coluna "PlayTimeDiff": sem registro quer dizer que não teve diferença de tempo naquela jogada

Tratei e agora que não vai ter mais nulo troquei a tipagem para ficar com um inteiro menor e economizar espaço

```python
df["down"] = df["down"].fillna(0).astype("int8")
df["FirstDown"] = df["FirstDown"].fillna(0).astype("int8")
df["GoalToGo"] = df["GoalToGo"].fillna(0).astype("int8")
df["PlayTimeDiff"] = df["PlayTimeDiff"].fillna(0).astype("int16")
```

#### Horário da jogada usando o "time"

Se o horário de uma jogada está faltando, vamos usar o horário da jogada anterior do mesmo jogo usando a variavel "GameID", assim não pegamos o horário de outra partida.

```python
print("Nulos em time antes:", df["time"].isna().sum())

df["time"] = df.groupby("GameID")["time"].ffill() #Pesquisei e vi o ffil serve para preencher valores nulos com os últimos valores não nulos

print("Nulos em time depois:", df["time"].isna().sum())
```
